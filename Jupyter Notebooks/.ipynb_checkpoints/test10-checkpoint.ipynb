```python
import pandas as pd=
import matplotlib.pyplot as plt
# matplotlib is the library for plotting 

df = pd.read_csv("monthly_sales.csv")
df.head()
```

```python
plt.plot(df["Month"], df["Total_Sales"])
# this plt.plot will be used for plotting the line plot 

plt.title("Monthly Total Sales")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.show()
```

```python
plt.bar(df["Month"], df["Online_Sales"])

plt.title("Monthly Online Sales")
plt.xlabel("Month")
plt.ylabel("Online Sales")
plt.show()
```

```python
plt.bar(df["Month"], df["Offline_Sales"], color="red")
plt.title("Monthly Offline Sales")
plt.xlabel("Month")
plt.ylabel("Offline Sales")

plt.show()
```

```python
plt.plot(df["Month"], df["Profit"], marker="o")

plt.title("Monthly Profit")
plt.xlabel("Month")
plt.ylabel("Profit")

plt.show()
```

```python
plt.plot(df["Month"], df["Online_Sales"], label="Online")
plt.plot(df["Month"], df["Offline_Sales"], label="Offline")

plt.title("Online vs Offline Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()

plt.show()
```

```python
plt.bar(df["Month"], df["Marketing_Spend"], color="green")

plt.title("Monthly Marketing Spend")
plt.xlabel("Month")
plt.ylabel("Marketing Spend")

plt.show()
```

```python
plt.plot(df["Month"], df["Customer_Count"], marker="*")

plt.title("Monthly Customer Count")
plt.xlabel("Month")
plt.ylabel("Number of Customers")

plt.show()
```

```python
plt.scatter(df["Marketing_Spend"], df["Total_Sales"] , marker="*")

plt.title("Marketing Spend vs Total Sales")
plt.xlabel("Marketing Spend")
plt.ylabel("Total Sales")

plt.show()
```

```python
plt.hist(df["Total_Sales"], bins=6, edgecolor="black" , orientation = "horizontal")

plt.title("Total Sales Distribution")
plt.xlabel("Total Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
sales = [
    df["Online_Sales"].sum(),
    df["Offline_Sales"].sum(),
    df["Total_Sales"].sum()
]

labels = ["Online Sales", "Offline Sales" , "Total Sale"]

plt.pie(sales, labels=labels, autopct="%1.1f%%")

plt.title("Online vs Offline Sales Contribution")

plt.show()
```

```python
monthly_sale = df["Total_Sales"]
months = df["Month"]

plt.pie(monthly_sale , labels= months , autopct="%1.1f%%")
plt.title("Monthwise sales distibution")
plt.show()
```

```python
monthly_sale = df["Profit"]
months = df["Month"]

plt.pie(monthly_sale , labels= months , autopct="%1.1f%%")
plt.title("Monthwise profit distibution".title())
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np


plt.bar(df["Month"], df["Total_Sales"], label="Income")
plt.bar(df["Month"], df["Marketing_Spend"], label="Spending")

plt.xlabel("Month")
plt.ylabel("Amount")
plt.title("Monthly Income vs Spending")
plt.legend()

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

corr = df.select_dtypes(include="number").corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```
