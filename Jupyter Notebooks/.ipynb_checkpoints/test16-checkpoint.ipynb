```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Lasso , Ridge
import pandas as pd
```

```python
df = pd.read_csv("polynomial_regression_dataset.csv")
poly = PolynomialFeatures(20)
```

```python
X = df[["Hours"]]
y = df["Marks"]
```

```python
x_poly = poly.fit_transform(X)
```

## Using model Linear regression

```python
model = LinearRegression()
```

```python
model.fit(x_poly , y)
```

```python
pred_res = model.predict(x_poly)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(df["Hours"] , df["Marks"])
plt.plot(df["Hours"] , pred_res)
```

## Using model Lasso

```python
model = Lasso(alpha=0.1)
model.fit(x_poly , y)
```

```python
model.coef_
```

```python
res = model.predict(x_poly)
```

```python
plt.plot(df["Hours"] , df["Marks"])
plt.scatter(df["Hours"] , res)
```

## Using model Ridge

```python
model = Ridge()
```

```python
model.fit(x_poly , y)
```

```python
model.coef_
```

```python
pred_res = model.predict(x_poly)
```

```python
plt.plot(df["Hours"] , df["Marks"])
plt.plot(df["Hours"] , pred_res)
```
