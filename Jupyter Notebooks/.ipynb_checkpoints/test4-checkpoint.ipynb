# Assignment: Handling Missing Values Using Mean, Median, and Mode

## Objective
Understand the mathematical calculation of Mean, Median, and Mode to fill missing values in a dataset.

### 1. Mean (Average)

The ages of five students are:

`20, 22, 24, 26, NaN`

**Tasks:**
1. Calculate the mean of the available ages.
2. Replace the missing value (`NaN`) with the calculated mean.
3. Write the final list of ages.

**Formula:**

Mean = Sum of all values / Number of values

### 2. Median (Middle Value)

The salaries of five employees are:

`15000, 20000, 25000, 30000, NaN`

**Tasks:**
1. Arrange the available salaries in ascending order.
2. Find the median salary.
3. Replace the missing value (`NaN`) with the calculated median.
4. Write the final list of salaries.

**Hint:** If the number of values is even, the median is the average of the two middle values.

### 3. Mode (Most Frequent Value)

The departments of employees are:

`IT, HR, IT, Sales, IT, HR, NaN`

**Tasks:**
1. Count the frequency of each department.
2. Identify the mode.
3. Replace the missing value (`NaN`) with the mode.
4. Write the final list of departments.

### 4. Bonus Questions

1. Calculate the mean of `10, 20, 30, 40, 50`.
2. Find the median of `12, 8, 15, 10, 20, 25`.
3. Find the median of `5, 10, 15, 20, 25`.
4. Find the mode of `2, 3, 3, 4, 5, 3, 6`.
5. If a dataset contains an extremely high salary, which is generally more suitable for filling a missing salary: Mean or Median? Explain why.

**Instructions:** Solve all questions using mathematical calculations only. Do not use Python, Pandas, or any built-in functions.
