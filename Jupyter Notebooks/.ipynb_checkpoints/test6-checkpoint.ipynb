```python
# again you want to do the change in existing column 
# Example: "New Package"
# multiple ways are there but for simplest thing just go with the apply method
# whats the apply method

# you just create the function which will take each record from the dataframe 
# and apply the operation on that value or 
# example company said that every emplpyees salary will be deducted by 1 lakh as the quater was very
# bad no revenue generation in this quater 

# if salary is greater than 9 lakhs then 2 lakh will be reduced and if less than 9 lakh then 1 lakh 
# reduction
```

```python
def salary_reducer(value):
    if value >= 900000:
        return value - 200000
    else:
        return value - 100000
```

```python
df["Salary Reduction"] = df["New Package"].apply(salary_reducer)
```

```python
df
```

```python
# what will happen in ml
# you cant pass the string to model for training 
# you should have the numbers only 
# then how will we convert the string into number
```

```python
def gender_to_number(gender):
    if gender == "Male":
        return 0
    if gender == "Female":
        return 1
```

```python
df["Gender"] = df["Gender"].apply(gender_to_number)
```

```python
df
```

```python
# hr = 1
# finance = 2
# it = 3

def dept_to_number(dept):
    if dept == "HR":
        return 1
    elif dept == "Sales":
        return 2
    elif dept == "IT":
        return 3
    elif dept == "Finance":
        return 4
    else:
        return 5
```

```python
df["Department"] = df["Department"].apply(dept_to_number)
```
