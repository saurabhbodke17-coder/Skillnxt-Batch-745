```python
import pandas as pd

employees = pd.DataFrame({
    "EmpID": [1, 2, 3, 4],
    "Name": ["Rahul", "Priya", "Amit", "Sneha"],
    "DeptID": [101, 102, 101, 104]
})

departments = pd.DataFrame({
    "DeptID": [101, 102, 103],
    "Department": ["IT", "HR", "Finance"]
})
```

```python
employees
```

```python
departments
```

```python
result = pd.merge(
    employees,
    departments,
    on="DeptID",
    how="inner"
)

result
```

```python
result = pd.merge(
    employees,
    departments,
    on="DeptID",
    how="left"
)

result
```

```python
result = pd.merge(
    employees,
    departments,
    on="DeptID",
    how="right"
)

result
```

```python
result = pd.merge(
    employees,
    departments,
    on="DeptID",
    how="outer"
)
result
```
