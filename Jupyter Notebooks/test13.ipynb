## Multiple Linear Regression

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

```python
df = pd.read_csv("company_employees.csv")
df.head()
```

```python
df1 = df[["Department" , "Age" , "Experience" , "PerformanceRating" , "ProjectsCompleted" , "Salary"]]
```

```python
df1
```

```python
df1["Department"].unique()
```

```python
def str_to_num(x):
    
    if x.lower() == "Sales".lower():
        return 1
        
    elif x.lower() == "Finance".lower():
        return 2

    elif x.lower() == "Marketing".lower():
        return 3

    elif x.lower() == "HR".lower():
        return 4

    elif x.lower() == "IT".lower():
        return 5
```

```python
df1["Department"] = df1["Department"].apply(str_to_num)
```

```python
df1.isnull().sum()
```

```python
df1
```

```python
x = df1[["Department",	"Age"	,"Experience",	"PerformanceRating"	,"ProjectsCompleted"]]
y= df1["Salary"]
```

```python
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.20)
```

```python
x_train
```

```python
y_train
```

```python
model = LinearRegression()
model.fit(x_train , y_train)
```

```python
x_test.shape
```

```python
preds = model.predict(x_test)
preds
```

```python
print(y_test)
```

```python
res = mean_squared_error(y_test , preds)
print(res)
```

```python
record_level_error = res ** 0.5
record_level_error
```
