```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("polynomial_regression_dataset.csv")
df
```

```python
plt.plot(df["Hours"] , df["Marks"])
```

```python
X = df[["Hours"]]
y = df["Marks"]
```

```python
poly_feture = PolynomialFeatures(degree=6)
```

```python
x_poly = poly_feture.fit_transform(X)
```

```python
x_poly
```

```python
model = LinearRegression()
```

```python
model.fit(x_poly , y)
```

```python
res = model.predict(x_poly)
res
```

```python
# now before using model.predict we will always transform the fetures using poly_feture

# my input is 5
user_input = poly_feture.transform([[78]])
model.predict(user_input)
```
