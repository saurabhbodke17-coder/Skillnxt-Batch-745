## Groupby

```python
df
```

```python
# these are the people who has max salary in every department 
df.groupby("Department")["Salary"].max()
```

```python
# minimum salary in every department 
df.groupby("Department")["Salary"].min()
```

```python
# average salary in every department 
df.groupby("Department")["Salary"].mean()
```

```python
# Department wise report 
df.groupby("Department")["Salary"].agg(["min" , "max" , "mean" , "count" , "sum"])
```

```python
df.groupby("City")["Salary"].agg(["min" , "max" , "mean" , "count" , "sum"])
```
